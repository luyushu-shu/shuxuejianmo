# 多元回归线性分析

## 引入头文件

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
```

## 设置中文格式和样式

```python
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
sns.set_style("whitegrid")
```

## # 1. 数据生成

```python
np.random.seed(42)
n_samples = 200
```

## 生成8个关键变量

```python
X_key = np.random.randn(n_samples, 8)
```

## 定义真实的回归系数（关键变量）

```python
true_coefficients = np.array([2.5, -1.8, 3.2, 1.5, -2.1, 2.8, -1.3, 1.9])
```

## 生成2个误差变量（与目标变量无关或负相关）

```python
X_noise = np.random.randn(n_samples, 2)
```

## 计算真实的目标变量

```python
y_true = np.dot(X_key, true_coefficients) + np.random.normal(0, 0.5, n_samples)
```

## 合并所有变量

```python
X_all = np.concatenate([X_key, X_noise], axis=1)
```

## 创建DataFrame

```python
feature_names = [f'关键变量_{i+1}' for i in range(8)] + [f'误差变量_{i+1}' for i in range(2)]
df = pd.DataFrame(X_all, columns=feature_names)
df['目标变量'] = y_true

print("数据集基本信息:")
print(f"样本数量: {n_samples}")
print(f"特征数量: {X_all.shape[1]} (8个关键变量 + 2个误差变量)")
print(f"目标变量统计: 均值={y_true.mean():.2f}, 标准差={y_true.std():.2f}")
print("\n数据集前5行:")
print(df.head())
```

## 2. 逐步添加关键变量的回归分析

```python
def progressive_regression_analysis(X, y, feature_names, title_prefix=""):
    """
    逐步添加变量进行回归分析
    """
    results = []
    predictions_list = []
    
    for i in range(1, X.shape[1] + 1):
        # 使用前i个变量
        X_subset = X[:, :i]
        
        # 拟合模型
        model = LinearRegression()
        model.fit(X_subset, y)
        
        # 预测
        y_pred = model.predict(X_subset)
        predictions_list.append(y_pred)
        
        # 计算评价指标
        mse = mean_squared_error(y, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y, y_pred)
        r2 = r2_score(y, y_pred)
        
        results.append({
            '变量数量': i,
            '使用变量': feature_names[:i],
            'MSE': mse,
            'RMSE': rmse,
            'MAE': mae,
            'R²': r2
        })
    
    return results, predictions_list
```

## 3. 关键变量逐步添加分析

```python
print("\n" + "="*50)
print("关键变量逐步添加分析")
print("="*50)

key_results, key_predictions = progressive_regression_analysis(
    X_key, y_true, [f'关键变量_{i+1}' for i in range(8)], "关键变量"
)

# 显示结果
key_results_df = pd.DataFrame(key_results)
print(key_results_df[['变量数量', 'MSE', 'RMSE', 'MAE', 'R²']].round(4))
```

## 4. 添加误差变量的分析

```python
print("\n" + "="*50)
print("添加误差变量后的分析")
print("="*50)

# 使用所有8个关键变量作为基础，然后逐步添加误差变量
X_base = X_key  # 8个关键变量
X_with_noise = np.concatenate([X_base, X_noise], axis=1)

noise_results, noise_predictions = progressive_regression_analysis(
    X_with_noise, y_true, feature_names, "包含误差变量"
)

# 只显示基础模型和添加误差变量后的结果
noise_results_df = pd.DataFrame(noise_results)
comparison_results = noise_results_df[noise_results_df['变量数量'].isin([8, 9, 10])]
print(comparison_results[['变量数量', 'MSE', 'RMSE', 'MAE', 'R²']].round(4))
```

## 5. 可视化分析

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
```

## 5.1 评价指标变化图

```python
# 5.1 评价指标变化图
ax1 = axes[0, 0]
key_metrics = key_results_df[['变量数量', 'MSE', 'R²']]
ax1_twin = ax1.twinx()

line1 = ax1.plot(key_metrics['变量数量'], key_metrics['MSE'], 'bo-', label='MSE', linewidth=2, markersize=6)
line2 = ax1_twin.plot(key_metrics['变量数量'], key_metrics['R²'], 'ro-', label='R²', linewidth=2, markersize=6)

ax1.set_xlabel('关键变量数量')
ax1.set_ylabel('MSE', color='blue')
ax1_twin.set_ylabel('R²', color='red')
ax1.set_title('关键变量逐步添加 - 评价指标变化')
ax1.grid(True, alpha=0.3)

# 合并图例
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax1_twin.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='center right')
```

## 5.2 预测vs真实值散点图（最终模型）

```python
ax2 = axes[0, 1]
final_pred_key = key_predictions[-1]
ax2.scatter(y_true, final_pred_key, alpha=0.6, color='green')
ax2.plot([y_true.min(), y_true.max()], [y_true.min(), y_true.max()], 'r--', linewidth=2)
ax2.set_xlabel('真实值')
ax2.set_ylabel('预测值')
ax2.set_title(f'8个关键变量模型预测效果\nR² = {key_results[-1]["R²"]:.4f}')
ax2.grid(True, alpha=0.3)
```

##  5.3 误差变量影响对比

```python
ax3 = axes[1, 0]
comparison_data = comparison_results[['变量数量', 'MSE', 'R²']]
ax3_twin = ax3.twinx()

x_pos = [0, 1, 2]
x_labels = ['8个关键变量', '+误差变量1', '+误差变量2']

bars1 = ax3.bar([x-0.2 for x in x_pos], comparison_data['MSE'], width=0.4, 
                label='MSE', color='lightcoral', alpha=0.7)
bars2 = ax3_twin.bar([x+0.2 for x in x_pos], comparison_data['R²'], width=0.4, 
                     label='R²', color='lightblue', alpha=0.7)

ax3.set_xlabel('模型配置')
ax3.set_ylabel('MSE', color='red')
ax3_twin.set_ylabel('R²', color='blue')
ax3.set_title('误差变量对模型性能的影响')
ax3.set_xticks(x_pos)
ax3.set_xticklabels(x_labels, rotation=15)

# 添加数值标签
for i, (bar1, bar2) in enumerate(zip(bars1, bars2)):
    height1 = bar1.get_height()
    height2 = bar2.get_height()
    ax3.text(bar1.get_x() + bar1.get_width()/2., height1 + 0.01,
             f'{height1:.3f}', ha='center', va='bottom', fontsize=8)
    ax3_twin.text(bar2.get_x() + bar2.get_width()/2., height2 + 0.01,
                  f'{height2:.3f}', ha='center', va='bottom', fontsize=8)

lines3, labels3 = ax3.get_legend_handles_labels()
lines4, labels4 = ax3_twin.get_legend_handles_labels()
ax3.legend(lines3 + lines4, labels3 + labels4, loc='center right')
```

## 5.4 残差分析

```python
ax4 = axes[1, 1]
residuals_key = y_true - final_pred_key
residuals_noise = y_true - noise_predictions[-1]

ax4.scatter(final_pred_key, residuals_key, alpha=0.6, label='仅关键变量', color='green')
ax4.scatter(noise_predictions[-1], residuals_noise, alpha=0.6, label='包含误差变量', color='red')
ax4.axhline(y=0, color='black', linestyle='--', alpha=0.8)
ax4.set_xlabel('预测值')
ax4.set_ylabel('残差')
ax4.set_title('残差分析对比')
ax4.legend()
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. 相关性分析热图

```python
plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))

sns.heatmap(correlation_matrix, mask=mask, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=0.5, cbar_kws={"shrink": 0.8}, fmt='.3f')
plt.title('变量相关性热图')
plt.tight_layout()
plt.show()
```

## 7. 创建动画可视化函数

```python
def create_progressive_animation():
    """
    创建关键变量逐步添加的动画效果
    """
    from matplotlib.animation import FuncAnimation
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    def animate(frame):
        ax1.clear()
        ax2.clear()
        
        # 当前使用的变量数量
        n_vars = frame + 1
        
        # 更新评价指标图
        current_metrics = key_results_df.iloc[:n_vars]
        ax1.plot(current_metrics['变量数量'], current_metrics['MSE'], 'bo-', 
                linewidth=2, markersize=8, label='MSE')
        ax1_twin = ax1.twinx()
        ax1_twin.plot(current_metrics['变量数量'], current_metrics['R²'], 'ro-', 
                     linewidth=2, markersize=8, label='R²')
        
        ax1.set_xlabel('关键变量数量')
        ax1.set_ylabel('MSE', color='blue')
        ax1_twin.set_ylabel('R²', color='red')
        ax1.set_title(f'关键变量逐步添加 (当前: {n_vars}个变量)')
        ax1.grid(True, alpha=0.3)
        ax1.set_xlim(0.5, 8.5)
        
        # 更新预测vs真实值图
        current_pred = key_predictions[n_vars-1]
        ax2.scatter(y_true, current_pred, alpha=0.6, color='green')
        ax2.plot([y_true.min(), y_true.max()], [y_true.min(), y_true.max()], 'r--', linewidth=2)
        ax2.set_xlabel('真实值')
        ax2.set_ylabel('预测值')
        ax2.set_title(f'预测效果 (R² = {key_results[n_vars-1]["R²"]:.4f})')
        ax2.grid(True, alpha=0.3)
        
    anim = FuncAnimation(fig, animate, frames=8, interval=1000, repeat=True)
    plt.tight_layout()
    return anim

def create_noise_animation():
    """
    创建误差变量影响的动画效果
    """
    from matplotlib.animation import FuncAnimation
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # 基础数据（8个关键变量的结果）
    base_result = noise_results[7]  # 8个关键变量
    
    def animate(frame):
        ax1.clear()
        ax2.clear()
        
        if frame == 0:
            # 只显示关键变量结果
            current_result = base_result
            title_suffix = "仅关键变量"
            color = 'green'
            current_pred = noise_predictions[7]
        elif frame == 1:
            # 添加第一个误差变量
            current_result = noise_results[8]
            title_suffix = "添加误差变量1"
            color = 'orange'
            current_pred = noise_predictions[8]
        else:
            # 添加第二个误差变量
            current_result = noise_results[9]
            title_suffix = "添加误差变量2"
            color = 'red'
            current_pred = noise_predictions[9]
        
        # 更新评价指标柱状图
        metrics = ['MSE', 'R²']
        values = [current_result['MSE'], current_result['R²']]
        bars = ax1.bar(metrics, values, color=[color, color], alpha=0.7)
        
        # 添加数值标签
        for bar, value in zip(bars, values):
            height = bar.get_height()
            ax1.text(bar.get_x() + bar.get_width()/2., height + height*0.01,
                    f'{value:.4f}', ha='center', va='bottom', fontsize=12, fontweight='bold')
        
        ax1.set_title(f'模型性能指标 - {title_suffix}')
        ax1.set_ylabel('指标值')
        ax1.grid(True, alpha=0.3)
        
        # 更新预测vs真实值图
        ax2.scatter(y_true, current_pred, alpha=0.6, color=color)
        ax2.plot([y_true.min(), y_true.max()], [y_true.min(), y_true.max()], 'r--', linewidth=2)
        ax2.set_xlabel('真实值')
        ax2.set_ylabel('预测值')
        ax2.set_title(f'预测效果 - {title_suffix}\nR² = {current_result["R²"]:.4f}')
        ax2.grid(True, alpha=0.3)
        
    anim = FuncAnimation(fig, animate, frames=3, interval=1500, repeat=True)
    plt.tight_layout()
    return anim
```

```python
# 显示动画（注意：在Jupyter中需要特殊设置才能显示动画）
print("\n" + "="*50)
print("动画可视化")
print("="*50)
print("正在创建动画...")

try:
    # 创建并显示第一个动画
    anim1 = create_progressive_animation()
    plt.show()
    
    # 创建并显示第二个动画
    anim2 = create_noise_animation()
    plt.show()
    
    print("动画创建完成！")
    print("注意：在某些环境中，动画可能需要额外的设置才能正常显示。")
    
except Exception as e:
    print(f"动画创建遇到问题: {e}")
    print("请确保matplotlib和相关依赖库已正确安装。")
```
